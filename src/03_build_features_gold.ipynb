```python
# Databricks notebook source
from pyspark.sql import functions as F
dbutils.widgets.text("catalog", "workspace")
dbutils.widgets.text("schema", "churn_dev")
catalog, schema = dbutils.widgets.get("catalog"), dbutils.widgets.get("schema")

gold = (spark.table(f"{catalog}.{schema}.silver_customers")
    .withColumn("total_spend", F.col("tenure_months") * F.col("monthly_charges"))
    .withColumn("calls_per_year", F.col("support_calls") / (F.col("tenure_months") / 12))
    .withColumn("is_month_to_month", (F.col("contract") == "month-to-month").cast("int")))

gold.write.mode("overwrite").saveAsTable(f"{catalog}.{schema}.gold_churn_features")
```
